```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from math import tan
scalar = 3.4
arr = np.array([3.3,4.4,5.6,6.1,7.3,8.8,9.2,10.9])

add_values = scalar + arr
print("Addition: ",add_values)

sub_values = scalar - arr
print("Subtraction: ",sub_values)

mul_values = scalar * arr
print("Multiplication: ",mul_values)

div_values = scalar / arr
print("Division: ",div_values)

sq_values = scalar ** arr
print("Square: ",sq_values)

abs_values = np.abs(arr)
print("Absolute: ",abs_values)

exp_values = np.exp(arr)
print("Exponential: ",exp_values)

log_values = np.log(arr)
print("Logarithm: ",log_values)

sin_values = np.sin(arr)
print("Sine: ",sin_values)

cos_values = np.cos(arr)
print("Cosine: ",cos_values)

tan_values= np.tan(arr)
print("Tan: ",tan_values)

asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
print("Inverse Sine: ",asin_values)

acos_values = np.arccos(np.clip(arr/np.max(arr), -1, 1))
print("Inverse Cosine: ",acos_values)

atan_values = np.arctan(arr)
print("Inverse Tangent: ",atan_values)

```
