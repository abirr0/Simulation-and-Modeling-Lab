```python
import numpy as np
from scipy import stats
```

```python
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])
alpha = 0.05

n1 = len(data1)
n2 = len(data2)

mean1 = np.mean(data1)
mean2 = np.mean(data2)
var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)
mean_diff = mean1 - mean2


df = n1 + n2 - 2
sp2 = ((n1 - 1) * var1 + (n2 - 1) * var2) / df
se_diff = np.sqrt(sp2 * (1 / n1 + 1 / n2))

t_stat = mean_diff / se_diff
p_val = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=df))


t_crit = stats.t.ppf(1 - alpha / 2, df=df)
ci_lower = mean_diff - t_crit * se_diff
ci_upper = mean_diff + t_crit * se_diff


print(f"Sample 1 Mean: {mean1:.4f}, Variance: {var1:.4f}")
print(f"Sample 2 Mean: {mean2:.4f}, Variance: {var2:.4f}")
print(f"Mean Difference: {mean_diff:.4f}")
print(f"Degrees of Freedom: {df}")
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_val:.6e}")
print(f"95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")
```
