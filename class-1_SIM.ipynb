```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

arr = np.array([1,2,3,4,5])

sin_values = np.sin(arr)
print("Sin Value" ,sin_values)

cos_values = np.cos(arr)
print("Cos Values",cos_values)

tan_values = np.tan(arr)
print ("Tan Value : " ,tan_values)

asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
print ("Reverse Sin Value", asin_values)

acos_values = np.arccos(np.clip(arr/np.max(arr), -1,1 ))
print("Reverse Cos Value :" ,acos_values)

atan_values = np.arctan(arr)
print("Reverse Tan Values : ", atan_values)


exp_values = np.exp(arr)
print("Exponential Values :" ,exp_values)

abs_values = np.abs(arr)
print("Absolute Values : ", abs_values)

sqrt_values = np.sqrt(arr)
print("Squer Values :" , sqrt_values)

rem_values = np.remainder(arr, 3)
print("Remainder values :", rem_values)

temp = np.array([1.5,1.1,1.9])

round_values = np.round(temp)
print("Round Values :", round_values)

floor_value = np.floor(temp)
print ("Floor Values : ", floor_value)

ceil_values = np.ceil(temp)
print("Ceiling Values :",ceil_values)
```

```python
from mpmath.libmp.libelefun import MAX_LOG_INT_CACHE
vec = np.array([3, 1, 4, 1, 5, 9, 2, 6])

max_value = np.max(vec)
print(" Max Value : ",max_value)

max_index = np.argmax(vec)
print ("Maxi Index : " ,max_index)

min_value = np.min(vec)
print(" Min Value : ",min_value)

min_index = np.argmin(vec)
print ("Min Index : " ,min_index)

vec_length = len(vec)
print("Vector Length ",vec_length)

sorted_a_vec =np.sort(vec)
print("Assending Order of Vector :",sorted_a_vec)

sorted_d_vec =np.sort(vec)[::-2]
print("Dissending Order of Vector: ",sorted_d_vec)

sum_values =np.sum(vec)
print("Sum of : ",sum_values)

prod_values = np.prod(vec)
print("Product of Vector: ",prod_values)

median_value = np.median(vec)
print("Median is : ",median_value)

mean_value = np.mean(vec)
print("Mean is: ",mean_value)

std_dev = np.std(vec)
print("Standerd Deviation : ",std_dev)

```
